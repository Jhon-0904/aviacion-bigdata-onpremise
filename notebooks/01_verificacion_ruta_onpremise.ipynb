```python
import os
import sys
import glob
import traceback

# Entorno Java, Hadoop y Spark
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-arm64"
os.environ["HADOOP_HOME"] = "/opt/hadoop"
os.environ["HADOOP_CONF_DIR"] = "/opt/hadoop/etc/hadoop"
os.environ["YARN_CONF_DIR"] = "/opt/hadoop/etc/hadoop"
os.environ["SPARK_HOME"] = "/opt/spark-4.0.4-bin-hadoop3"

# Python del driver y de los workers
os.environ["PYSPARK_DRIVER_PYTHON"] = "/home/max/venv-jupyter/bin/python"
os.environ["PYSPARK_PYTHON"] = "/usr/bin/python3"

# Librerías PySpark y Py4J
sys.path.insert(
    0,
    "/opt/spark-4.0.4-bin-hadoop3/python/lib/pyspark.zip"
)

sys.path.insert(
    0,
    "/opt/spark-4.0.4-bin-hadoop3/python/lib/py4j-0.10.9.9-src.zip"
)

print("Iniciando Spark sobre YARN...")

try:
    from pyspark.sql import SparkSession

    spark = (
        SparkSession.builder
        .appName("VerificacionRutaAOnPremiseLATAM")
        .master("yarn")
        .config("spark.submit.deployMode", "client")
        .config("spark.pyspark.python", "/usr/bin/python3")
        .config(
            "spark.pyspark.driver.python",
            "/home/max/venv-jupyter/bin/python"
        )
        .config("spark.sql.shuffle.partitions", "48")
        .getOrCreate()
    )

    print("Spark creado correctamente")
    print("Versión Spark:", spark.version)
    print("Master:", spark.sparkContext.master)

except Exception:
    print("ERROR REAL AL CREAR SPARK:")
    traceback.print_exc()
```

```python
ruta_integrado = "hdfs:///proyecto_latam/processed/integrado"

df = spark.read.parquet(ruta_integrado)

print("Registros integrados:", df.count())
print("Columnas:", len(df.columns))

df.select(
    "Origin",
    "Dest",
    "FlightDate",
    "delay_target"
).show(10, truncate=False)
```

```python
spark.stop()
print("Verificación Ruta A completada correctamente.")
```

```python


from pyspark.sql import SparkSession

try:
    spark.stop()
except:
    pass

spark = (
    SparkSession.builder
    .appName("LecturaParquetRutaA")
    .master("yarn")
    .config("spark.submit.deployMode", "client")
    .config("spark.sql.shuffle.partitions", "48")
    .getOrCreate()
)

print("Spark activo:", spark.version)
print("Master:", spark.sparkContext.master)
```

```python
ruta_integrado = "hdfs:///proyecto_latam/processed/integrado"

df = spark.read.parquet(ruta_integrado)

print("Registros integrados:", df.count())
print("Columnas:", len(df.columns))

df.select(
    "Origin",
    "Dest",
    "FlightDate",
    "delay_target"
).show(10, truncate=False)
```

```python
df.groupBy("delay_target").count().orderBy("delay_target").show()
```

```python
spark.stop()
print("Verificación de la Ruta A on-premise completada correctamente.")
```
